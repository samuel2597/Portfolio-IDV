# Data Cleaning and Preprocessing with Pandas

Data cleaning is a crucial step in the data science workflow. In this tutorial, we'll learn how to handle missing values, deal with outliers, and convert categorical data to one-hot encoded data using pandas.

First, we import the necessary libraries.

```python
import pandas as pd
import numpy as np
import warnings
warnings.simplefilter('ignore')
```

## Sample Data

We'll create a sample DataFrame to work with:

```python
# Sample data
data = {
    'Age': [25, 27, 29, None, 23, np.nan, 30],
    'Salary': [50000, 54000, 50000, 45000, None, 52000, 58000],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', None],
    'Department': ['IT', 'HR', 'HR', 'IT', 'HR', 'IT', 'Finance']
}

df = pd.DataFrame(data)
df
```

## Imputing Missing Values

Missing values can be imputed in various ways depending on the nature of the data and the missingness.

### Numeric Columns

For numeric columns like 'Age' and 'Salary', we can impute missing values with the mean or median:

```python
# Fill missing values for 'Age' with the mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill missing values for 'Salary' with the median
df['Salary'].fillna(df['Salary'].median(), inplace=True)
df
```

### Categorical Columns

For categorical columns, we can impute missing values with the mode (the most frequent value):

```python
# Fill missing values for 'Gender' with the mode
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)

# For simplicity, let's assume 'Department' has no missing values
df
```

## Dealing with Outliers

Outliers can skew your data and affect the outcome of your analysis. Here's a simple method to detect and remove outliers based on the Z-score:

```python
from scipy import stats

# Calculate Z-scores of 'Age' and 'Salary'
z_scores = stats.zscore(df[['Age', 'Salary']])

# Define a threshold
threshold = 3

# Find where the absolute Z-scores are less than the threshold
without_outliers = (np.abs(z_scores) < threshold).all(axis=1)

# Filter the dataframe
df_cleaned = df[without_outliers]
df_cleaned
```

## Converting Categorical Data to One-Hot Encoded Data

One-hot encoding is a process by which categorical variables are converted into a form that could be provided to ML algorithms to do a better job in prediction.

```python
# One-hot encode 'Gender' and 'Department' columns
df_encoded = pd.get_dummies(df_cleaned, columns=['Gender', 'Department'])
df_encoded
```
